# Sistema de gestión de inventario (POO)

Aplicación de consola en Python orientada a objetos para gestionar el inventario de una tienda.

**Funcionalidades:**
- Alta de productos con validación de datos (precio y cantidad)
- Búsqueda de productos por nombre
- Listado completo del inventario
- Cálculo del valor total del inventario

**Conceptos aplicados:** clases y encapsulación (`Producto`, `Inventario`), manejo de excepciones (`ValueError`), validación de entradas de usuario, menú interactivo con bucle de control.

**Tecnologías:** Python (estándar, sin librerías externas)

La celda de código incluye una sesión de ejemplo ya ejecutada, mostrando el menú y su uso real.


```python
class Producto:
    '''
    Clase para los productos del inventario.
    
    Atributos:
        nombre (str): Nombre del producto.
        precio (float): Precio del producto (debe ser >= 0).
        cantidad (float): Cantidad disponible del producto (debe ser >= 0).
    '''
    def __init__(self, nombre, precio, cantidad):
        '''
        Método constructor de la clase. Inicializa un objeto de clase Producto

        Argumentos:
            nombre (str): Nombre del producto.
            precio (float): Precio del producto (debe ser >= 0).
            cantidad (float): Cantidad disponible del producto (debe ser >= 0).

        Excepciones:
            ValueError: Si los argumentos no son válidos.
        '''
         # Validar y asignar nombre
        if isinstance(nombre, str) and nombre.strip():
            self.nombre = nombre.strip()
        else:
            raise ValueError("El nombre debe ser una cadena no vacía.")
        # Validar y asignar precio
        self.actualizar_precio(precio)
        
        # Validar y asignar cantidad
        self.actualizar_cantidad(cantidad)
            
    def actualizar_precio(self, nuevo_precio): 
        '''
        Actualiza el precio del producto.
        
        Argumentos:
            nuevo_precio (float): Nuevo precio del producto (debe ser >= 0).

        Excepciones:
            ValueError: Si el nuevo precio no es válido.
        '''
        try:
            nuevo_precio_float = float(nuevo_precio)
        except (ValueError, TypeError):
            raise ValueError("El precio debe ser un número válido.")
        if nuevo_precio_float < 0:
            raise ValueError("El precio debe ser un número mayor o igual a 0.")
        self.precio = nuevo_precio_float

    def actualizar_cantidad(self, nueva_cantidad):
        '''
        Actualiza la cantidad del producto.
        
        Argumentos:
            nueva_cantidad (int): Nueva cantidad del producto (debe ser >= 0).

        Excepciones:
            ValueError: Si la nueva cantidad no es válido.
        '''
        try:
            nueva_cantidad_int = int(nueva_cantidad)
        except (ValueError, TypeError):
            raise ValueError("La cantidad debe ser un número válido.")
        if nueva_cantidad_int < 0:
            raise ValueError("La cantidad debe ser un número entero mayor o igual a 0.")
        self.cantidad = nueva_cantidad_int
            
    def calcular_valor_total(self):
        '''
        Método que calcula el valor total (precio × cantidad).

        Salida:
            float: Valor total del producto
        '''
        return self.precio*self.cantidad

    def __str__(self): 
        '''
        Método que muestra la información del producto de forma legible.

        Salida:
            str: Cadena con la información del producto.
        '''
        return(
            f"{'--------' * 6}\n"
            f"NOMBRE DEL PRODUCTO: -----> {self.nombre}\n"
            f"PRECIO DEL PRODUCTO: -----> {self.precio}\n"
            f"CANTIDAD DEL PRODUCTO: -----> {self.cantidad}\n"
            f"{'--------' * 6}\n"
            f"VALOR TOTAL DEL PRODUCTO: -----> {self.calcular_valor_total()}\n"
            f"{'--------' * 6}"
        )

class Inventario:
    '''
    Clase para almacenar y registrar los productos del inventario.
    
    '''

    def __init__(self):
        '''
        Método constructor del inventario. Inicializa un objeto de clase Inventario.

        Atributos:
            list: Lista vacía para almacenar productos.
        '''
        self.productos = [];

    def agregar_producto(self, producto):
        '''
        Añade un objeto de tipo Producto a la lista del inventario.

        Argumentos:
            Producto: Objeto de clase Producto
        '''
        productos = self.productos
        productos.append(producto)
        self.productos = productos

    def buscar_producto(self, nombre): 
        '''
        Búsqueda para encontrar un producto por su nombre (búsqueda exacta, insensible a mayúsculas/minúsculas).

        Argumentos:
            str: Nombre exacto del producto del inventario.

        Salida:
            str: Información del producto.

        Execciones: 
            Si el producto no existe devuelve None.
        '''
        for producto in self.productos:
            if nombre.strip().lower() == producto.nombre.lower():
                return producto.__str__()
        
        return None

    def calcular_valor_inventario(self):
        '''
        Calculo de el valor total del inventario.

        Salida:
            float: Suma total de todos los valores de los productos del inventario.
        '''
        valor = sum(producto.calcular_valor_total() for producto in self.productos)
        print(f"VALOR TOTAL DEL INVENTARIO: -----> {valor}\n")

    def listar_productos(self):
        '''
        Método que muestra la información de todos los productos del inventario de forma legible.

        Salida:
            str: Cadena con la información del inventario.
        '''
        
        resultado = f"{'========'*6}\nINVENTARIO:\n"
        for producto in self.productos:
            resultado += f"{producto.__str__()}\n"
        return resultado
        
            
def menu_principal():
    '''
    Muestra al usuario las opciones de acciones a realizar y permite salir de la opción después de realizar cada acción.

    Opciones:
        1. Agregar producto
        2. Buscar producto
        3. Listar productos
        4. Calcular valor total del inventario
        5. Salir
    '''
    while True:
        print("\n=== MENÚ PRINCIPAL ===")
        print("1. Agregar producto")
        print("2. Buscar producto")
        print("3. Listar productos")
        print("4. Calcular valor total del inventario")
        print("5. Salir")

        accion = input("\nIndique la acción que desea realizar pulsando su número correspondiente: ")
        try:
            accion_int = int(accion)
        except ValueError:
            print('El valor indicado es erroneo. Intentelo de nuevo.')
        else:
            
            # AGREGAR PRODUCTOS
            if accion_int == 1:
                print(f"\n{'===' *4} AGREGAR PRODUCTO {'===' *4}")
                nombre = input("Introduzca el nombre del producto que desea agregar: ")
                precio = input("Introduzca el precio del producto: ")
                cantidad = input("Introduzca la cantidad del producto: ")

                try:
                    producto = Producto(nombre, precio, cantidad)
                    inventario.agregar_producto(producto)
                    print(f"\nProducto '{producto.nombre}' agregado correctamente.")
                except ValueError as e:
                    print(f"Error al agregar el producto: {e}")
                    
            # BUSCAR PRODUCTOS
            elif accion_int == 2:
                print(f"\n{'===' *4} BUSCAR PRODUCTO {'===' *4}")
                nombre = input("Introduzca el nombre del producto que desea buscar:")
                
                if inventario.buscar_producto(nombre):
                    producto = inventario.buscar_producto(nombre)
                    print(f"\nProducto '{nombre}' encontrado correctamente.")
                    print(producto)
                else: 
                    print(f"\nProducto '{nombre}' no encontrado en el inventario.")
                    
            # LISTAR PRODUCTOS
            elif accion_int == 3:
                print(f"\n{'===' *4} LISTAR PRODUCTOS {'===' *4}")
                productos = inventario.listar_productos()
                print(productos)
                
            # CALCULAR VALOR TOTAL DEL INVENTARIO
            elif accion_int == 4:
                print(f"\n{'===' *4} CALULAR VALOR TOTAL DEL INVENTARIO {'===' *4}")
                inventario.calcular_valor_inventario()
                
            # SALIR
            elif accion_int == 5:
                print(f"\n{'===' *4} SALIENDO DEL PROGRAMA {'===' *4}")
                break

            # OPCIONES NO VÁLIDAS
            else:
                print("Opción no válida. Por favor seleccione una acción del 1 al 5.")

if __name__ == "__main__":
    inventario = Inventario()
    menu_principal()
```
